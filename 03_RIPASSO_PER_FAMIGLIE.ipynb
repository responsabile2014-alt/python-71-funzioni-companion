### Esercizio 2 - Collegamento tra più voci

**Obiettivo**

Trasformare e poi selezionare.

**Funzioni suggerite**

map(), filter()

**Consegna**

Da ["1", "2", "3", "4"] ricava gli interi e tieni soltanto i pari.

**PROVA TU**

```python
testi = ["1", "2", "3", "4"]
# Completa qui
```

**Suggerimento**

Scorri il risultato di map con un ciclo for, poi usa filter con una funzione definita con def.

**Soluzione**

```python
def pari(n):
    return n % 2 == 0

testi = ["1", "2", "3", "4"]
for numero in map(int, testi):
    print(numero)

for numero in filter(pari, map(int, testi)):
    print("pari:", numero)
```

**Spiegazione breve**

map trasforma tutti gli elementi, filter ne seleziona alcuni.

### Esercizio 3 - Controllo di comprensione

**Obiettivo**

Distinguere metodo di classe, metodo statico e property.

**Funzioni suggerite**

classmethod(), staticmethod(), property(), super(), object()

**Consegna**

Crea una classe Base e una sottoclasse che usa super(), con un classmethod, uno staticmethod e una property.

**PROVA TU**

```python
# Completa qui
```

**Suggerimento**

cls per il classmethod, nessun parametro implicito per lo staticmethod.

**Soluzione**

```python
class Base:
    def __init__(self, nome):
        self.nome = nome

class Prodotto(Base):
    def __init__(self, nome, prezzo):
        super().__init__(nome)
        self.prezzo = prezzo

    @property
    def con_iva(self):
        return self.prezzo * 1.22

    @classmethod
    def gratuito(cls, nome):
        return cls(nome, 0)

    @staticmethod
    def descrizione():
        return "Prodotto in catalogo"

p = Prodotto("Penna", 10)
print(p.con_iva)
print(Prodotto.gratuito("Omaggio").prezzo)
print(Prodotto.descrizione())
print(vars(p))
```

**Spiegazione breve**

La property si legge senza parentesi; classmethod riceve la classe; staticmethod non riceve nulla.
