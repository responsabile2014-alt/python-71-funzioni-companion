### SCENARIO

### FUNZIONI CHE ALLENI

map(), filter(), int(), list(), sum(), print()

### LA SFIDA

Parti da:

testi = ["10", "-3", "25", "0", "8"]

Converti tutti i testi in interi.

Poi conserva soltanto i numeri maggiori di zero e calcolane il totale.

### PROVA TU

````python
testi = ["10", "-3", "25", "0", "8"]

# Completa qui la soluzione
````

### OUTPUT ATTESO

```
Numeri: [10, -3, 25, 0, 8]
Positivi: [10, 25, 8]
Totale positivi: 43
```

### SOLUZIONE

````python
def positivo(numero):
    return numero > 0

testi = ["10", "-3", "25", "0", "8"]

numeri = list(map(int, testi))
positivi = list(filter(positivo, numeri))

print("Numeri:", numeri)
print("Positivi:", positivi)
print("Totale positivi:", sum(positivi))
````

### SCENARIO

Hai i nomi di quattro partecipanti e i relativi punteggi. Vuoi costruire la classifica dal punteggio più alto al più basso.

### FUNZIONI CHE ALLENI

zip(), sorted(), enumerate(), print()

### LA SFIDA

Usa:

nomi = ["Anna", "Luca", "Sara", "Marco"]
punti = [82, 95, 88, 91]

Abbina i punteggi ai nomi, ordina la classifica e numerala partendo da 1.

### PROVA TU

````python
nomi = ["Anna", "Luca", "Sara", "Marco"]
punti = [82, 95, 88, 91]

# Completa qui la soluzione
````

### OUTPUT ATTESO

```

2 Marco 91
3 Sara 88
4 Anna 82
```

### SOLUZIONE

````python
nomi = ["Anna", "Luca", "Sara", "Marco"]
punti = [82, 95, 88, 91]

classifica = sorted(zip(punti, nomi), reverse=True)

for posizione, (punteggio, nome) in enumerate(classifica, 1):
    print(posizione, nome, punteggio)
````

## Laboratorio 6 - Controllare se un modulo è completo

### SCENARIO

Hai quattro campi di un modulo. Uno è vuoto. Vuoi sapere se tutti sono compilati e se esiste almeno un campo mancante.

### FUNZIONI CHE ALLENI

bool(), all(), any(), len(), print()

### LA SFIDA

Parti da:

campi = ["Anna", "Rossi", "anna@email.it", ""]

Crea un controllo True/False per ogni campo e verifica:

- se tutti sono compilati;

- se almeno uno è vuoto;

- quanti campi esistono in totale.

### PROVA TU

````python
campi = ["Anna", "Rossi", "anna@email.it", ""]

# Completa qui la soluzione
````

### OUTPUT ATTESO

```
Tutti i campi compilati: False
C'è almeno un campo vuoto: True
Campi totali: 4
```

### SOLUZIONE

````python
campi = ["Anna", "Rossi", "anna@email.it", ""]

compilati = []
vuoti = []

for valore in campi:
    compilato = bool(valore)
    compilati.append(compilato)
    vuoti.append(not compilato)

print("Tutti i campi compilati:", all(compilati))
print("C'è almeno un campo vuoto:", any(vuoti))
print("Campi totali:", len(campi))
````

## Laboratorio 10 - Creare il riepilogo di un ordine

### SCENARIO

Hai tre liste parallele: prodotti, prezzi e quantità. Vuoi costruire le righe dell’ordine e calcolare il totale.

### FUNZIONI CHE ALLENI

zip(), enumerate(), list(), sum(), format(), print()

### LA SFIDA

Parti da:

prodotti = ["Penna", "Quaderno", "Zaino"]
prezzi = [1.5, 3.2, 24.9]
quantita = [3, 2, 1]

Per ogni prodotto calcola il subtotale, cioè prezzo per quantità. Mostra poi le righe numerate e il totale finale.

### PROVA TU

````python
prodotti = ["Penna", "Quaderno", "Zaino"]
prezzi = [1.5, 3.2, 24.9]
quantita = [3, 2, 1]

# Completa qui la soluzione
````

### OUTPUT ATTESO

```
1 Penna 3 4.50
2 Quaderno 2 6.40
3 Zaino 1 24.90
Totale ordine: 35.80
```

### SOLUZIONE

````python
prodotti = ["Penna", "Quaderno", "Zaino"]
prezzi = [1.5, 3.2, 24.9]
quantita = [3, 2, 1]

righe = list()
subtotali = list()

for nome, prezzo, qta in zip(prodotti, prezzi, quantita):
    subtotale = prezzo * qta
    righe.append((nome, qta, subtotale))
    subtotali.append(subtotale)

for numero, (nome, qta, subtotale) in enumerate(righe, 1):
    print(numero, nome, qta, format(subtotale, ".2f"))

print("Totale ordine:", format(sum(subtotali), ".2f"))
````
